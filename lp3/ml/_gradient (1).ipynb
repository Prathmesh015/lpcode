```python

# Define the function
def function(x):
    return (x + 3) ** 2

# Define the gradient (derivative)
def gradient(x):
    return 2 * (x + 3)

# Starting point
x = 2

# Learning rate
learning_rate = 0.1

# Number of iterations
iterations = 50

# Apply Gradient Descent
for i in range(iterations):
    x = x - learning_rate * gradient(x)

# Display the result
print("Local minimum point (x):", round(x, 6))
print("Minimum value (y):", round(function(x), 6))
```
